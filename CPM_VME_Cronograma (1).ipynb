### Video 5: Programación del Proyecto (Lista de Actividades y Secuencia Lógica)

**Definición de Actividades, Atributos e Hitos**

La programación del proyecto parte de la descomposición jerárquica de los paquetes de trabajo de la EDT/WBS (nivel más bajo del alcance) en actividades individuales programables. Este proceso, llamado *definición de actividades*, produce unidades de trabajo lo suficientemente pequeñas como para estimar su duración y asignarles recursos con precisión.

Cada actividad se documenta mediante sus **atributos**, que típicamente incluyen: identificador único, nombre y descripción, código de la EDT al que pertenece, predecesoras y sucesoras, tipo de relación lógica, adelantos/retrasos, requisitos de recursos, restricciones y supuestos, y el responsable de ejecutarla. Estos atributos son los que después alimentan el cálculo de la ruta crítica.

Las actividades se conectan además a **hitos**: puntos o eventos clave del proyecto (por ejemplo, "Arquitectura aprobada" o "Ambiente de producción disponible") que tienen **duración cero** y sirven como puntos de control para medir avance frente a la línea base.

**Método de Diagramación de Precedencia (PDM)**

Es la técnica estándar para construir el modelo de cronograma: representa cada actividad como un nodo y las dependencias entre actividades como flechas dirigidas (diagrama de red tipo actividad-en-el-nodo). Existen cuatro tipos de relación lógica:

- **Final a Inicio (FS):** la sucesora no puede iniciar hasta que finalice la predecesora. Es la relación más frecuente en proyectos de software (p. ej., "Codificación" no puede iniciar hasta que termine "Diseño de Arquitectura").
- **Inicio a Inicio (SS):** la sucesora inicia cuando inicia la predecesora (ambas pueden avanzar en paralelo desde el mismo punto de arranque, p. ej., "Documentación técnica" inicia junto con "Desarrollo").
- **Final a Final (FF):** la sucesora finaliza cuando finaliza la predecesora (p. ej., "Pruebas de regresión" debe cerrar al mismo tiempo que "Corrección de errores").
- **Inicio a Final (SF):** la predecesora finaliza cuando inicia la sucesora; es la relación menos usada y aparece sobre todo en esquemas de transición o reemplazo de sistemas (el sistema legado se apaga cuando arranca el nuevo).

**Adelantos (Leads) y Retrasos (Lags)**

Son modificadores temporales que se aplican sobre la relación lógica para ajustar el cronograma sin cambiar la dependencia en sí:

- **Adelanto (Lead):** permite solapar actividades adelantando el inicio de la sucesora antes de que la predecesora termine del todo (se modela como un valor negativo, p. ej., "Pruebas unitarias" puede iniciar 3 días antes de que termine por completo "Codificación de Módulos").
- **Retraso (Lag):** impone una espera obligatoria entre el fin de la predecesora y el inicio de la sucesora (valor positivo, p. ej., "Despliegue" espera 2 días después de "Aprobación de cambios" por ventana de mantenimiento).

**Entregables de la sesión:** la *Lista de Actividades con sus Atributos* y el *Diagrama de Red del Cronograma del Proyecto*, que son la entrada directa para el cálculo de la ruta crítica.

### Audio 5: Ingeniería de Cronogramas y Método de la Ruta Crítica (CPM)

**Método de la Ruta Crítica (CPM)**

Una vez construida la red de precedencias, el Método de la Ruta Crítica identifica la secuencia continua de actividades que representa el **camino más largo** a través de la red. Esa secuencia determina la **duración mínima total** en la que puede completarse el proyecto: cualquier retraso en una actividad de la ruta crítica retrasa la fecha de finalización del proyecto en la misma cantidad de días, mientras que las actividades fuera de la ruta crítica tienen margen de maniobra (holgura).

**Cálculo de fechas — los dos pases del algoritmo CPM**

- **Pase Hacia Adelante (Forward Pass):** recorre la red desde el inicio del proyecto hacia el final, calculando para cada actividad el **Inicio Temprano (ES)** y la **Finalización Temprana (EF)** — las fechas más tempranas posibles en que puede empezar y terminar, respetando sus predecesoras. Fórmula: `EF = ES + Duración − 1`.
- **Pase Hacia Atrás (Backward Pass):** recorre la red desde el final del proyecto hacia el inicio, calculando la **Finalización Tardía (LF)** y el **Inicio Tardío (LS)** — las fechas más tardías en que una actividad puede terminar/iniciar sin retrasar el proyecto. Fórmula: `LS = LF − Duración + 1`.

**Análisis de holguras**

- **Holgura Total (Total Float):** el margen máximo que una actividad puede retrasarse sin posponer la fecha de finalización del proyecto (`LS − ES` o, equivalentemente, `LF − EF`). Las actividades de la ruta crítica tienen holgura total igual a **cero**; identificar qué actividades tienen `TF = 0` es, de hecho, la forma práctica de encontrar la ruta crítica.
- **Holgura Libre (Free Float):** el margen que una actividad puede retrasarse sin afectar el Inicio Temprano (ES) de su actividad sucesora inmediata. Es siempre menor o igual a la holgura total, y es especialmente útil cuando varias actividades convergen en una misma sucesora (como ocurre en el caso práctico de este notebook).

**Costo de Demora en la Ruta Crítica (Cost of Delay)**

Es el impacto financiero directo — o la reducción del Retorno de la Inversión (ROI) — que produce cada día de retraso en una actividad situada sobre la ruta crítica. Este concepto es el puente hacia el análisis cuantitativo de riesgos: al asignarle un valor monetario a cada día de atraso, es posible traducir la incertidumbre del cronograma (retrasos probables en actividades críticas o casi críticas) en cifras financieras mediante el **Valor Monetario Esperado (VME)**, que se desarrolla en la sección 3 de este notebook.

```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
```

## 1. Actividades del proyecto (EDT descompuesta)

```python
actividades = {
    "A": {"nombre": "Diseño de Arquitectura y Base de Datos", "duracion": 10, "predecesoras": []},
    "B": {"nombre": "Desarrollo de Módulos Core / Backend",    "duracion": 15, "predecesoras": ["A"]},
    "C": {"nombre": "Integración de API Externa de Pagos",     "duracion": 12, "predecesoras": ["A"]},
    "D": {"nombre": "Desarrollo de Interfaz Frontend",         "duracion": 10, "predecesoras": ["A"]},
    "E": {"nombre": "Pruebas Integradas y QA",                 "duracion": 8,  "predecesoras": ["B", "C", "D"]},
    "F": {"nombre": "Despliegue en Cloud y Salida a Producción","duracion": 5,  "predecesoras": ["E"]},
}

df_act = pd.DataFrame(actividades).T
df_act.index.name = "ID"
df_act
```

## 2. Método de la Ruta Crítica (CPM)

Se construye la red de precedencias (todas las relaciones son Final-Inicio, FS) y se resuelve el pase hacia adelante y el pase hacia atrás.

```python
G = nx.DiGraph()
for act_id, datos in actividades.items():
    G.add_node(act_id, **datos)
    for pred in datos["predecesoras"]:
        G.add_edge(pred, act_id)

orden = list(nx.topological_sort(G))

# --- Pase hacia adelante (ES, EF) ---
ES, EF = {}, {}
for act_id in orden:
    dur = actividades[act_id]["duracion"]
    preds = actividades[act_id]["predecesoras"]
    ES[act_id] = 1 if not preds else max(EF[p] for p in preds) + 1
    EF[act_id] = ES[act_id] + dur - 1

duracion_total = max(EF.values())

# --- Pase hacia atrás (LF, LS) ---
LF, LS = {}, {}
sucesores = {n: list(G.successors(n)) for n in G.nodes}
for act_id in reversed(orden):
    dur = actividades[act_id]["duracion"]
    sucs = sucesores[act_id]
    LF[act_id] = duracion_total if not sucs else min(LS[s] for s in sucs) - 1
    LS[act_id] = LF[act_id] - dur + 1

# --- Holguras ---
TF = {a: LS[a] - ES[a] for a in orden}          # Holgura total
FF = {}                                          # Holgura libre
for act_id in orden:
    sucs = sucesores[act_id]
    FF[act_id] = (min(ES[s] for s in sucs) - EF[act_id] - 1) if sucs else TF[act_id]

df_cpm = pd.DataFrame({
    "Actividad": [actividades[a]["nombre"] for a in orden],
    "Duración": [actividades[a]["duracion"] for a in orden],
    "ES": [ES[a] for a in orden], "EF": [EF[a] for a in orden],
    "LS": [LS[a] for a in orden], "LF": [LF[a] for a in orden],
    "Holgura Total": [TF[a] for a in orden],
    "Holgura Libre": [FF[a] for a in orden],
    "¿Crítica?": ["Sí" if TF[a] == 0 else "No" for a in orden],
}, index=orden)
df_cpm.index.name = "ID"
df_cpm
```

```python
ruta_critica = [a for a in orden if TF[a] == 0]
print(f"Duración total del proyecto: {duracion_total} días")
print(f"Ruta crítica: {' → '.join(ruta_critica)}")
```

### Diagrama de red (ruta crítica resaltada)

```python
pos = nx.spring_layout(G, seed=7, k=1.2)
colores = ["#e74c3c" if n in ruta_critica else "#95a5a6" for n in G.nodes]

plt.figure(figsize=(9, 5))
nx.draw_networkx_nodes(G, pos, node_size=1600, node_color=colores)
nx.draw_networkx_labels(G, pos, font_color="white", font_weight="bold")
nx.draw_networkx_edges(G, pos, arrowsize=20, width=1.5)
plt.title("Diagrama de red del cronograma — rojo = ruta crítica")
plt.axis("off")
plt.tight_layout()
plt.show()
```

### Diagrama de Gantt (ruta crítica resaltada)

```python
fig, ax = plt.subplots(figsize=(9, 4))
for i, a in enumerate(orden):
    color = "#e74c3c" if a in ruta_critica else "#3498db"
    ax.barh(a, EF[a] - ES[a] + 1, left=ES[a] - 1, color=color)
    if TF[a] > 0:
        ax.barh(a, TF[a], left=EF[a], color="#bdc3c7", alpha=0.6)

ax.set_xlabel("Día del proyecto")
ax.set_title("Cronograma (Gantt) — rojo = crítica, gris = holgura")
ax.legend(handles=[mpatches.Patch(color="#e74c3c", label="Actividad crítica"),
                    mpatches.Patch(color="#3498db", label="Actividad con holgura"),
                    mpatches.Patch(color="#bdc3c7", label="Holgura disponible")],
          loc="lower right")
plt.tight_layout()
plt.show()
```

## 3. Cuantificación de riesgos del cronograma — Valor Monetario Esperado (VME)

`VME = Probabilidad × Impacto financiero`. Si el retraso de una actividad excede su holgura total, el excedente se convierte en atraso real del proyecto (y por tanto en costo de demora).

```python
costo_demora_dia = 1500  # USD por día de retraso en la fecha de finalización del proyecto

riesgos = {
    "R1": {"actividad": "C", "descripcion": "Integración API Externa de Pagos",
           "probabilidad": 0.40, "retraso_dias": 6},
    "R2": {"actividad": "E", "descripcion": "Incompatibilidad Backend/QA",
           "probabilidad": 0.25, "retraso_dias": 4},
}

filas = []
for rid, r in riesgos.items():
    act = r["actividad"]
    holgura = TF[act]
    retraso_proyecto = max(0, r["retraso_dias"] - holgura)
    impacto_financiero = retraso_proyecto * costo_demora_dia
    vme = r["probabilidad"] * impacto_financiero
    filas.append({
        "Riesgo": rid, "Actividad": act, "Descripción": r["descripcion"],
        "Probabilidad": r["probabilidad"], "Retraso actividad (días)": r["retraso_dias"],
        "Holgura disponible (días)": holgura, "Retraso al proyecto (días)": retraso_proyecto,
        "Impacto financiero (USD)": impacto_financiero, "VME (USD)": vme,
    })

df_vme = pd.DataFrame(filas).set_index("Riesgo")
df_vme
```

```python
reserva_contingencia = df_vme["VME (USD)"].sum()
print(f"Reserva de contingencia recomendada: USD {reserva_contingencia:,.2f}")
```

## 4. Conclusiones

- La ruta crítica del proyecto es **A → B → E → F**, con una duración total de **38 días**. C y D tienen holgura (3 y 5 días respectivamente) porque corren en paralelo a B, que es la más larga de las tres.
- **R1 (Actividad C)** tiene 3 días de holgura disponibles frente a un retraso posible de 6 días: solo 3 días de ese retraso llegan a afectar la fecha final del proyecto.
- **R2 (Actividad E)** está sobre la ruta crítica (holgura = 0), así que cualquier retraso ahí impacta el proyecto día por día.
- La reserva de contingencia (suma de los VME) es el monto que debería incorporarse a la línea base de costos para cubrir estos dos riesgos de cronograma.
