## Resumen de clase del 15-09-2026

### Video 7: Ruta Crítica y Compresión

El video explica cómo analizar y organizar las actividades de un proyecto de software para determinar el tiempo mínimo necesario para completarlo y cómo reducir su duración cuando sea necesario.

Los temas principales son el Método de la Ruta Crítica (CPM), el cálculo de fechas tempranas y tardías mediante los recorridos hacia adelante y hacia atrás, y el análisis de holguras. La ruta crítica está formada por las actividades que determinan la duración total del proyecto y normalmente tienen holgura total igual a cero.

También se explica el costo de demora, relacionado con las pérdidas económicas o sobrecostos que pueden producirse por retrasar actividades importantes. Finalmente, se estudian dos técnicas de compresión del cronograma: **Crashing**, que consiste en agregar recursos, personal u horas extras, y **Fast-Tracking**, que consiste en ejecutar actividades en paralelo. Esta última puede aumentar el riesgo de errores y retrabajos.

El entregable principal es un diagrama de Gantt con la ruta crítica identificada y un plan de contingencia.

### Video 8: Dominio de Recursos

El video explica cómo planificar, organizar y administrar los recursos necesarios para ejecutar un proyecto de software de manera eficiente.

Se estudia la planificación, estimación y adquisición de recursos, así como la diferencia entre recursos humanos, físicos y virtuales. Los recursos humanos incluyen desarrolladores, líderes de proyecto y demás integrantes del equipo; los recursos físicos y virtuales incluyen equipos, licencias, herramientas digitales e infraestructura en la nube.

También se presenta la **Estructura de Desglose de Recursos (RBS)**, que permite organizar los recursos por categorías y subcategorías. Otro tema importante es la **nivelación de recursos**, utilizada para detectar sobreasignaciones, distribuir mejor la carga de trabajo y ajustar el cronograma cuando sea necesario.

Los entregables de esta sesión son la Estructura de Desglose de Recursos (RBS) y el calendario de recursos del proyecto.

### Conclusión

Los videos muestran la importancia de administrar correctamente el tiempo y los recursos en un proyecto de software. La ruta crítica permite identificar las actividades que afectan la duración del proyecto, mientras que la compresión ayuda a reducir los tiempos. Por su parte, el dominio de recursos permite organizar y distribuir adecuadamente el equipo, las herramientas y la infraestructura necesarios para cumplir los objetivos del proyecto.
