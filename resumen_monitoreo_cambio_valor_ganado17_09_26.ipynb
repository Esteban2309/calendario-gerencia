## 11. Monitoreo y Cambio

El video **11 Monitoreo y Cambio** explica cómo supervisar y controlar un proyecto de software, especialmente el cronograma y las finanzas. El objetivo es comparar continuamente el avance real del proyecto con la línea base aprobada para detectar desviaciones y tomar decisiones oportunas.

### Temas principales
- **Monitoreo del cronograma y las finanzas:** permite revisar si el proyecto avanza según lo planeado y si los costos se mantienen dentro del presupuesto.
- **Recolección de datos de desempeño laboral:** consiste en registrar el avance real, los recursos utilizados y las métricas del proyecto.
- **Gestión de variaciones:** se analizan las diferencias entre lo planeado y lo ejecutado para identificar retrasos, sobrecostos o problemas de alcance.
- **Control de cambios:** las modificaciones deben evaluarse formalmente antes de ser aplicadas.
- **Comité de Control de Cambios (CCB):** revisa las solicitudes de cambio y puede aprobarlas, aplazarlas o rechazarlas según su impacto en el alcance, tiempo y costo.

## 12. Gestión del Valor Ganado (EVM)

El video **12 Gestión del Valor Ganado** presenta la metodología **Earned Value Management (EVM)**, utilizada para medir el desempeño de un proyecto mediante la comparación entre el trabajo planeado, el trabajo realizado y el costo real. Esta técnica ayuda a determinar si el proyecto está adelantado, retrasado, dentro del presupuesto o generando sobrecostos.

### Dimensiones principales
- **Valor Planificado (PV):** presupuesto autorizado para el trabajo que debía realizarse hasta una fecha determinada.
- **Costo Real (AC):** costo que realmente se ha gastado para ejecutar el trabajo realizado.
- **Valor Ganado (EV):** valor presupuestado del trabajo que realmente se ha completado.

### Análisis de variaciones

**Variación del Costo (CV):**

$$CV = EV - AC$$

- Si CV es positivo, existe un ahorro o el proyecto está por debajo del presupuesto.
- Si CV es negativo, existe un sobrecosto.

**Variación del Cronograma (SV):**

$$SV = EV - PV$$

- Si SV es positivo, el proyecto está adelantado respecto al trabajo planificado.
- Si SV es negativo, el proyecto presenta retraso.

### Aplicación práctica
La metodología EVM se aplica en una fecha de corte o **Status Date**. En ese momento se calculan PV, AC y EV, y luego se interpretan CV y SV para conocer el estado del proyecto y tomar decisiones correctivas.

### Entregable
El entregable principal es un **Reporte de Desempeño del Valor Ganado**, que reúne las mediciones de costo y tiempo para apoyar la toma de decisiones.
